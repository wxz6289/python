```python
from itemloaders.processors import Join, MapCompose
from os.path import join

# Join()(["Hi", "King"])
s = MapCompose(str.strip)([u' I ', u' am \n'])
print(s)

s = MapCompose(str.strip, str.title)([u' I ', u' am \n'])
print(s)

print(MapCompose(float)(["3.12"]))
print(MapCompose(lambda s: s.replace(',', ''),float)(["111,30.12"]))
mc = MapCompose(lambda u: join('http://my.com/test/abc', u))
print(mc(['example.html#check']))
```
