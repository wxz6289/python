```python
from langchain_classic import LLMChain
from langchain_classic.prompts import ChatPromptTemplate
from langchain_classic.llms import OpenAI

llm = OpenAI(
    model="deepseek-chat", base_url="http://api.deepseek.com/v1", temperature=0
)

prompt = "请帮为{product}写一个广告语"

chain = LLMChain(llm=llm, prompt=ChatPromptTemplate.from_template(prompt), verbose=True)

chain("幸惠超市")
```
