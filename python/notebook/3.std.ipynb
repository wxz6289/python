## argparse

```python
import argparse

parser = argparse.ArgumentParser(
    prog="top", description="show top lines from each file "
)
parser.add_argument("filenames", nargs="+")
parser.add_argument("-l", "--lines", type=int, default=10)
args = parser.parse_args()
print(args)
```

## re

```python
import re

pattern = r"\bf[a-z]*"

re.findall(pattern, "which foot and handle feel fastest")
```

```python
re.sub(r"(\b[a-z]+) \1", r"\1", "cat in the the hat")
```

## random

```python
import random

random.choice(range(1, 10))
random.sample(range(100), 10)
print(random.random())
print(random.randrange(100))
```

## statistics

```python
import statistics

data = [1.2, 2.3, 2.1, 0.9, 5.6, 3.8]
statistics.mean(data)
statistics.median(data)
print(statistics.variance(data))
```

```python
from urllib.request import urlopen

with urlopen("https://docs.python.org/zh-cn/3/tutorial/stdlib.html") as response:
    for line in response:
        line = line.decode()
        if line.find("互联网访问"):
            print(line.rstrip())
```

## reprlib
repr() 缩略显示大型或深层嵌套的容器对象

```python
import reprlib

reprlib.repr(set("absbbsbsbbsjjdjdjksjdaifragilisticexpialidociou"))
```

pprint 模块提供了更加复杂的打印控制

```python
import pprint
t = [[[['black', 'cyan'], 'white', ['green', 'red']], [['magenta',
    'yellow'], 'blue']]]

pprint.pprint(t, width=30)
```

```python
import textwrap
doc = """The wrap() method is just like fill() except that it returns
a list of strings instead of one big string with newlines to separate
the wrapped lines."""

print(textwrap.fill(doc, width=40))
```
