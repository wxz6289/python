# 文件操作
- 打开文件 open()
- 文件模式
  - r 只读模式
  - w 写入模式
  - x 独占写模式
  - a 附加模式
  - t 文本模式
  - `+` 读写模式
  - r+与w+的区别 w+会删除内容
- 读取 read()/readline()
- 写入 write()
- seek(offset [, whence]) 将读写位置移动到offset位置 
  whence 的取值
  - io.SEEK_SET(0) 相对文件开头 默认值
  - is.SEEK_CUR(1) 相对当前位置 
  - io.SEEK_END(2) 相对文件结尾
- tell() 当前位置
- flush() 将缓冲区内容写入文件
- close() 关闭文件

```python
f = open("./test/text.txt", "w+")
f.write("Hello")
f.write("King")

# print(f)
f.close()
```

## 区分目录和文件

```python
from os import listdir, path

dirs = []
files = []
for df in listdir(".."):
    d = "../" + df
    if path.isfile(d):
        files.append(d)
    elif path.isdir(d):
        dirs.append(d)
    else:
        print(d)

print(f"dirs: {dirs}")
print(f"files: {files}")
```

```python
from os import listdir, remove, rename
from os.path import basename, join, isfile, isdir


def remove_exe(dir):
    if isdir(dir):
        for df in listdir(dir):
            fp = join(dir, df)
            if isfile(fp) and df[-3:] == "exe":
                remove(fp)


def rename_file(dir):
    if isdir(dir):
        for df in listdir(dir):
            fp = join(dir, df)
            if isfile(fp) and df[-3:] == "dmg":
                new_name = "2024-" + df
                rename(fp, join(dir, new_name))

rename_file("/Users/dreamerking/Downloads")
# remove_exe("/Users/dreamerking/Downloads")
```

```python
with open("./test/text.txt") as f:
    count = 0
    char = f.read(1)
    while char:
        count += 1
        char = f.read(1)
    print("count: ", count)
```

```python
with open("./test/text.txt") as f:
    count = 0
    while True:
        char = f.read(1)
        if not char:
            break
        count += 1
    print("count: ", count)
```

```python
with open("./test/text.txt") as f:
    lines = 0
    while True:
        line = f.readline()
        if not line:
            break
        lines += 1
    print("line number: ", lines)
```

## shelve 
python数据存储到文件

```python
import shelve

# l = [p for p in dir(shelve) if not p.startswith('_')]
# print(l)
# help(shelve.open)
s = shelve.open("./test/test.dat")
s["x"] = ["a", "b", "c"]
s["x"].append("d")
print(s["x"])
s.close()
```

```python
import shelve

s = shelve.open("./test/test.dat", writeback=True)
s["x"] = ["a", "b", "c"]
s["x"].append("d")
print(s["x"])
s.close()
```

```python
import shelve

with shelve.open("./test/test.dat") as s:
    s["x"] = ["a", "b", "c"]
    temp = s["x"]
    temp.append("d")
    s["x"] = temp
    print(s["x"])
    # print(dir(s))
    for [k, v] in s.items():
        print(k, v)

    s.clear()
```

## json/pickle
- dumps() 数据类型转换成字符串
- dump() 数据类型转换成字符串并保存入文件
- loads() 字符串转换成数据类型
- load() 读取文件并转换成数据类型

json可以在不同语言之间交换数据, pickle只在python中使用
json只能序列化基本类型，如列表、字典、字符串、数值等，不能序列化日期、类对象等
pickle可以序列化所有类型，包括类、函数等


```python
import json

# data = json.load("./test/test.json")

# l = [p for p in dir(json) if not p.startswith('_')]
# print(l)

with open("./test/test.json", "r+") as json_file:
    content = json_file.read()
    json_content = json.loads(content)
    print(content)
    print(json_content)
    print(json_content["name"])
```

```python
import json

with open("./test/test.json", "r+") as json_file:
    content = json.load(json_file)
    print(content)
    print(content["name"])
    content["phone"] = "15658001296"
    address = ["浙江省", "杭州市"]
    content["address"] = address
    with open("./test/test.json", "w") as w_file:
        json.dump(content, w_file, ensure_ascii=False)
```
