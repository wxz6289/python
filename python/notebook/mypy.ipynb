+ Any 默认类型，一种魔法类型，渐进式类型系统的基础，位于类型层次结构的顶部和底部，允许所有可能的操作。与任意类型相容。
+ 任何类型都是object类型的字类型。越泛化的类型支持的操作越少。

`|`运算符可以用于isinstance和issubclass的第二个参数

协议即接口，指定需要实现的一个或多个方法的类型，供类型检查工具进行检查是否实现了特定方法。
实现协议的类不会与定义协议的类建立任何关系，不继承，也不用注册。类型检查工具负责查找可用的协议类型，实行用法检查。

Callable 用于注解回调函数或高阶函数返回的可调用对象
Callable[[ParamType1, ParamType2], ReturnType]

可选或关键字类型没有专门的注解句法

NoReturn 仅用于注解不返回函数的返回值类型，通常会抛出异常。

类型不完美，测试须全面

类型提示并不能取代自动化测试

静态类型很难发现的问题
- 误报
- 漏报
- 一些便利功能无法做静态检查,如拆包
- 类型检查工具对特性、描述符、元类和元编程等高级功能支持很差，或者根本无法理解
- 类型检查工具滞后于Python版本变化
- 常见的数据约束在类型系统中无法表达，类型提示对捕获业务逻辑错误没有帮助。
