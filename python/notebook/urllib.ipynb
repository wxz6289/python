# urllib、urllib2、urllib3

urllib3 提供了HTTP客户端功能的第三方库
request 提供了便利的请求处理功能

```python
from urllib.request import urlopen
import re

webpage = urlopen("http://python.org")
content = webpage.read()
m = re.search(b'<a href="([^"]+)" .*?>about</a>', content, re.IGNORECASE)
print(m.group(1))
```

```python
from urllib.request import urlretrieve

urlretrieve("http://python.org", "./python.html")
```

urlretrive() 下载文件到指定位置或临时目录
urlcleanup() 清理下载到临时目录的文件
quote(string [, safe]) 编码字符串为合法的URL
unquote(string) 解码
urlencode(query [, doseq]) 将映射或由包含两个元素的元组组成的序列转换成URL编码的字符串

```python
from urllib.request import quote, unquote

qs = quote("你好 hello ? - * ab")
print(qs)
print(unquote(qs))
```

```python
import urllib3

http = urllib3.PoolManager()
url = "http://www.baidu.com"
response = http.request("GET", url)
# print(response.data.decode("utf-8"))
```

```python
import requests

url = "http://www.google.com"
response = requests.get(url)
# print(response.text)
```
