# 异常

错误是异常的常见情况。异常对象表示异常状态，遇到错误时引发异常，异常未被处理(或捕获)时，程序会报错并退出。

else子句 try块中没有引发异常时执行
finally 无论try块中是否引发异常都执行
在函数中引发异常时，异常将传播到调用函数的地方。
警告类似于异常，通常指打印警告信息。可以指定警告类别。

```python
try:
    1 / 0
except ZeroDivisionError:
    print("oops!")
    raise
```

## 异常处理方式
- try...except
- try...except...else
- try...except...finally

使用try/except捕获异常。异常捕获后，如果要继续向上传播，可以调用raise且不提供任何参数(也可以显式提供捕获到的异常)重新引发它。

```python
try:
    x = input("Enter fist number: x = ")
    y = input("Enter fist number: y = ")
    x = int(x)
    y = int(y)
    z = x / y
    print(f"{x} / {y} = {z}")
except ZeroDivisionError:
    print(f"y = {y}")
    print(f"y can't be zero!")
```

```python
class MuffledCalculate:
    muffled = False

    def calc(self, expr):
        try:
            return eval(expr)
        except ZeroDivisionError:
            if self.muffled:
                print("Division by zero is illegal.")
            else:
                raise


calculator = MuffledCalculate()
result = calculator.calc("10 / 2")
print(result)

calculator.muffled = True
result = calculator.calc("10 / 0")
print(result)
```

```python
import sys

try:
    x = input("Enter fist number: x = ")
    # sys.exit()
    y = input("Enter fist number: y = ")
    x = int(x)
    y = int(y)
    z = x / y
    print(f"{x} / {y} = {z}")
# 捕获所有异常,不指定任何异常类型 这种捕获所有异常的方式不可取，不仅会隐藏预期的错误，还会隐藏没有考虑到的错误。甚至还捕获用户使用Ctrl+C终止执行的企图、调用sys.exit()来终止执行的企图。
# 大多数情况下，最好使用`except Exception as e`并对异常对象进行检查,将不是从Exception派生出来的异常排除，如SystemExit和KeyboardInterrupt(从BaseException派生而来)。
except:
    print("oops!")
```

### try...except...else
else块 没有异常时执行  
finally块 无论是否有异常都执行

```python
try:
    print("A simple task")
except:
    print("What? something went wrong.")
else:
    print("Ah ... It went planed.")
finally:
    print("always running...")
```

```python
while True:
    try:
        x = input("Enter fist number: x = ")
        y = input("Enter fist number: y = ")
        x = int(x)
        y = int(y)
        z = x / y
        print(f"{x} / {y} = {z}")
    except:
        print("Invalid input, Please try again")
    else:
        break
```

```python
while True:
    try:
        x = input("Enter fist number: x = ")
        y = input("Enter fist number: y = ")
        x = int(x)
        y = int(y)
        z = x / y
        print(f"{x} / {y} = {z}")
    except Exception as e:
        print("Invalid input", e)
        print("Please try again")
    else:
        break
```

```python
x = None
try:
    x = 1 / 0
except (NameError, ZeroDivisionError):
    print("Unknown available")
else:
    print("That went well")
finally:
    print("Clean up ...")
    del x
```

```python
def faulty():
    raise Exception("Something is wrong!")


def ignore_exception():
    faulty()


def handle_exception():
    try:
        faulty()
    except:
        print("Exception handled.")


# ignore_exception()
handle_exception()
```

```python
def describe_person(person):
    print("Description of ", person["name"])
    print("Age:", person["age"])
    if "occupation" in person:
        print("Occupation:", person["occupation"])


def describe_person2(person):
    print("Description of ", person["name"])
    print("Age:", person["age"])
    try:
        print("Occupation:", person["occupation"])
    except KeyError:
        pass


person = {
    "name": "King",
    "age": 20,
    # "occupation": "Play"
}

describe_person(person)
describe_person2(person)
```

```python
class B(Exception):
    pass


class C(B):
    pass


class D(C):
    pass


for cls in [B, C, D]:
    try:
        raise cls()
    except D:
        print("D")
    except C:
        print("C")
    except B:
        print("B")


for cls in [B, C, D]:
    try:
        raise cls()
    except B:
        print("B")
    except D:
        print("D")
    except C:
        print("C")
```

```python
try:
    raise Exception("spam", "eggs")
except Exception as inst:
    print(type(inst))  # 异常的类型
    print(inst.args)  # 参数保存在 .args 中
    print(inst)  # __str__ 允许 args 被直接打印，
    # 但可能在异常子类中被覆盖
    x, y = inst.args  # 解包 args
    print("x =", x)
    print("y =", y)
```
