```python
# re
通配符 `.` 除换行符之外的字符
转义特殊字符 `\` 解释器执行的转义和模块re执行的转义
字符集 `[]` 一个字符 排除指定字符集 在字符集前加`^` 连续区间范围使用`-`连接 在字符集中, `.`、`*`、`?`、`+`等特殊字符无需转义(转义也合法)。`^`字符位于字符集开头时需要转义。`]`和`-`字符不位于字符集开头时需要转义。
二选一 `|`
子模式 `()`
可选模式 `?`
重复模式 `*`、`+`、`{n,m}`
锚定 `^`、`$` 放在模式开头或结尾
```

```python
s1 = "python\\.org"
s2 = r"python\.org"
print(s1)
print(s2)
```

re模块中的重要函数
- compile(pattern [, flags]) 创建模式对象
- search(pattern, string [, flags]) 字符串中查找模式匹配
- match(pattern, string [, flags]) 在字符串开头匹配模式
- split(pattern, string) 根据模式分割字符串
- findall(pattern, string) 返回字符串中所有与模式匹配的子串列表
- sub(pat, repl, string [, count = 0]) 替换字符串中匹配的模式的子串
- escape(string) 转义字符串中对正则有特殊含义的字符

search、match等函数调用，如果提供字符串表示的正则表达式，都必须在内部转换成模式对象。模式对象上也有搜索和匹配等方法。

```python
from re import search, compile, match

purl = r"(https?://)?(w{3}\.)?python\.org"
py = compile(purl)
pys = "http://python.org"
result = search(purl, pys)
result2 = search(py, pys)
result3 = py.search(pys)
print(result)
print(result2)
print(result3)

print(match("p", "wp"))
print(match("p", "pw"))
```

```python
from re import split, compile

txt = "alpha, beta, ,,, gamma delta"
sep = compile("[, ]+")
print(split(sep, txt))
print(sep.split(txt))
print(sep.split(txt, maxsplit=2))
print(sep.split(txt, maxsplit=1))
```

```python
from re import findall

pat = "[a-zA-Z]+"
text = (
    "Everything negative - pressure, challenges - is all an opportunity for me to rise."
)
print(findall(pat, text))
```

```python
from re import sub

pat = "{name}"
text = "Dear {name} ..."
print(sub(pat, "King", text))
```

```python
from re import escape, match

text = "https://www.python.org"
pat = escape(text)
print(pat)
print(match(pat, text))
```

## 编组 MatchObject对象
- group([grop1, ...])
- start([group])
- end([group])
- span([group])

```python
from re import match

pat = "There (was a (wee) (cooper)) who (live in Fyfe)"
text = "There was a wee cooper who live in Fyfe"
result = match(pat, text)
print(result.group())
print(result.group(1))
print(result.span(2))
print(text[12:15])
```

```python
from re import sub

emphasis_pattern = r"\*([^*]+)\*"
text = "* who are you? *"
print(sub(emphasis_pattern, r"<em>\1</em>", text))
```

```python
from re import sub, compile, VERBOSE
import re

emphasis_pattern = r"""
\*      #起始一个*
([^*]+) #除了*之外的字符
\*      #结束一个*
"""

compiled_pattern = compile(emphasis_pattern, re.VERBOSE)
print(compiled_pattern.sub(r"<em>\1</em>", "Hello *python*"))
```

```python
from re import sub, compile, VERBOSE
import re

emphasis_pattern = r"""
([*]+)   #起始一个*
(.+?)   #使用非贪婪匹配
\1    #结束一个*
"""

compiled_pattern = compile(emphasis_pattern, re.VERBOSE)
print(compiled_pattern.sub(r"<em>\2</em>", "Hello ***python**** This is *it*!"))
```

```python
from re import compile
from fileinput import input

pat = compile("From: (.*) <.*?>$")

for line in input():
    m = pat.match(line)
    if m:
        print(m.group(1))
```

```python
import re, fileinput

filed_pattern = re.compile(r"\[(.+?)\]")

scope = {}

def replacement(match):
  code = match.group(1)
  try:
    return str(eval(code, scope))
  except SyntaxError:
    exec(code, scope)
    return ''

lines = []
  for line in fileinput.input():
    lines.append(line)

text = ''.join(lines)

print(filed_pattern.sub(replacement, text))
```
